```python
import socket

# Step 1: Create a socket
mysock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

# Step 2: Connect to icio.us (if it's still resolvable)
server = ('icio.us', 80)  # Try DNS resolution
try:
    mysock.connect(server)
except socket.gaierror:
    print("Error: Could not resolve icio.us. The domain may no longer exist.")
    exit()

# Step 3: Send HTTP GET request
cmd = 'GET / HTTP/1.1\r\nHost: icio.us\r\nConnection: close\r\n\r\n'
mysock.send(cmd.encode())

# Step 4: Receive and print the response
while True:
    data = mysock.recv(512)
    if not data:
        break
    print(data.decode(), end='')

# Step 5: Close the socket
mysock.close()
```

```python
import urllib.request
import string
from collections import Counter

# Open and read the URL
url = 'http://textfiles.com/stories/antcrick.txt'
response = urllib.request.urlopen(url)

# Initialize lists
all_words = []
cleaned_words = []

# Basic stopwords list
stopwords = set([
    "the", "and", "is", "to", "a", "of", "in", "it", "that", "was", "as",
    "for", "with", "on", "he", "she", "at", "by", "an", "be", "this", "from", "or"
])

# Process each line
for line in response:
    decoded_line = line.decode().strip().lower()                    # Decode and lowercase
    words = decoded_line.split()                                    # Split into words
    all_words.extend(words)                                         # Count all words (before cleaning)
    
    # Remove punctuation and stopwords
    cleaned_line = decoded_line.translate(str.maketrans('', '', string.punctuation))
    cleaned = [word for word in cleaned_line.split() if word not in stopwords]
    cleaned_words.extend(cleaned)

# Print results
print("Total words before cleaning:", len(all_words))
print("Total words after cleaning :", len(cleaned_words))

# Optional: Most frequent cleaned words
print("\n Top 10 most frequent cleaned words:")
word_freq = Counter(cleaned_words)
for word, count in word_freq.most_common(10):
    print(f"{word}: {count}")
```
